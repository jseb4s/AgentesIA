```python
from langchain_core.prompts import PromptTemplate


template = """\
# Instrucciones para el Asistente de IA

Recuerda que hoy es {fecha_actual}

## Rol
Eres un experto en marketing digital con 10 años de experiencia.

## Tarea
Analiza el siguiento texto de un anuncio y proporciona 3 sugerencias para mejorarlo, enfocandote en el llamado a la acción.

## Texto del anuncio
{texto_anuncio}

## Formato de salida
- *Sugerencia 1:* [Tu primera sugerencia]
- *Sugerencia 2:* [Tu segunda sugerencia]
- *Sugerencia 3:* [Tu tercera sugerencia]

"""

prompt_template = PromptTemplate.from_template(template)
prompt = prompt_template.format(
    fecha_actual="2024-10-01",
    texto_anuncio="Compra nuestro producto y mejora tu vida hoy mismo!"
)
print(prompt)
```

### Que pasa si no le envio ninguna variable?


```python
prompt_template = PromptTemplate.from_template(template)
prompt = prompt_template.format()
```

Se genera error, pues se espera que se reciban las variables.


### La anterior situacion se soluciona con variables parciales, que son como variables por defecto a menos que se nombren explicitamente.

```python
import datetime
today = datetime.datetime.now().strftime("%Y-%m-%d")


prompt_template = PromptTemplate.from_template(template, partial_variables={"fecha_actual": today})
prompt = prompt_template.format(
    texto_anuncio="Compra nuestro producto y mejora tu vida hoy mismo!"
)
print(prompt)
```

### Langchain soporta templates avanzados, mas alla de los strings, y uno de los mejores es **Jinja**

Jinja es un sistema de templates que usa Django. Permite trabajar con bucles, condicionales, etc dentro del template.

- Instalacion: uv add Jinja2

## Prompting usando Jinja

```python
# Jinja facilita la creacion de templates mas complejos y dinamicos.
```

```python
template = """\
You are a helpful assistant that responds to the user's message.

{% if name %}
The user's name is {{ name }} and you can call them by that name.
{% endif -%}

The user's message is: "{{ user_message }}"
"""


prompt_template = PromptTemplate.from_template(template, template_format="jinja2")
prompt = prompt_template.format(user_message="Como puedo mejorar mi pagina web?")
print(prompt)
```
