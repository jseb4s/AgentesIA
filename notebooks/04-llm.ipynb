### Historial de forma manual

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

msg_1 = SystemMessage(content="You are a helpful assistant who talks in spanish.")
msg_2 = HumanMessage(content="Vivo en Colombia")
msg_3 = AIMessage(content="Hola, asi que eres de Colombia, saludos parcero!")
msg_4 = HumanMessage(content="En que pais vivo?")

history = [msg_1, msg_2, msg_3, msg_4]
response = gemini_llm.invoke(history)
response.pretty_print()
```

### Funcion agnostica para el uso de modelos (OpenAI, Claude, Google, etc)

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(model = "gemini-2.5-flash", 
                      model_provider="google_genai",
                      temperature=0
                      )

response = llm.invoke(history)

response.pretty_print()
```
