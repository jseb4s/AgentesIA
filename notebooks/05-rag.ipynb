```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("openai:gpt-4o", temperature=0)
```

```python
file_search_tool = {
    "type": "file_search",
    "vector_store_ids": ["vs_376gdhd772129hhh92"] # Este id se obtiene al crear el vector store en la web deOpenAi
}

llm_with_tools = llm.bind_tools([file_search_tool])
```

```python
response = llm_with_tools.invoke("Como mejoro el rendimiento de una pagina web?")
response.text
```

```python
response = llm_with_tools.invoke("Hola como estas?")
response.text
```
